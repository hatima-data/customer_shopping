```python
import pandas as pd
df=pd.read_csv('d:\customer_shopping.csv')
df
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df['Size']=df['Size'].fillna('Standard')
df['Size'].isnull().sum()
```

```python
df['Delivery Speed']=df['Delivery Speed'].fillna('Standard')
```

```python
df['Festival/Sale']=df['Festival/Sale'].fillna(0)
df['Online Store']=df['Online Store'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
top_category=df.groupby('Category')['Purchase Amount (₹)'].sum()
display(top_category)
```

```python
top_category=df.groupby('Category')['Purchase Amount (₹)'].agg(
    total_revenue='sum',
    avg_purchase='mean',
    transaction_count='count'
).sort_values(by='total_revenue')
display(top_category)
```

```python
#Age filter
df['age_group']=pd.cut(
    df['Age'],
    bins=[0,25,40,60,100],
    labels=['18-25', '26-40', '41-60', '60+']
)

age_analysis=df.groupby('age_group', observed=False)['Purchase Amount (₹)'].agg(
    total_spend='sum',
    avg_spent='mean'
)
display(age_analysis)
```

```python
correlatiom_matrix=df[['Discount (%)', 'Purchase Amount (₹)', 'Quantity']].corr()
display(correlatiom_matrix)
```

```python
payment_analysis=df.groupby('Payment Method')['Purchase Amount (₹)'].agg(
    total_revenue='sum',
    transaction_count='count'
).sort_values(by='total_revenue', ascending=False)
display(payment_analysis)
```
